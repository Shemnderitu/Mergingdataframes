```python
import pandas as pd
```

```python
df = [{'Fellowship_ID':1001,'FirstName':'Frodo','Skills':'Hiding'},{'Fellowship_ID':1002,'FirstName':'Samwise','Skills':'Gardening'},{'Fellowship_ID':1003,'FirstName':'Gandalf','Skills':'Spells'},{'Fellowship_ID':1004,'FirstName':'Pippin','Skills':'Fireworks'}]
df
```

```python
df1 =pd.DataFrame(df)
df1
```

```python
df = [{'Fellowship_ID':1001,'FirstName':'Frodo','Age':50},{'Fellowship_ID':1002,'FirstName':'Samwise','Age':39},{'Fellowship_ID':1006,'FirstName':'Legolas','Age':2931},{'Fellowship_ID':1007,'FirstName':'Elrond','Age':6520},{'Fellowship_ID':1008,'FirstName':'Barromir','Age':51}]
df2 = pd.DataFrame(df)
df2
```

```python
#inner join
df1.merge(df2)
```

```python
df1.merge(df2, on='Fellowship_ID', how='inner')
```

```python
df1.merge(df2, how = 'left')
```

```python
df1.merge(df2, how = 'right')
```

```python
df1.merge(df2, how = 'outer')
```

```python
#Join works best with indexes
df3 = df1.set_index('Fellowship_ID')
df4 = df2.set_index('Fellowship_ID')
df3.join(df4, how = 'outer', lsuffix= '_left', rsuffix= '_right')
```

```python
#Concatenate
pd.concat([df1, df2])
```

```python
pd.concat([df1, df2], axis = 1)
```
